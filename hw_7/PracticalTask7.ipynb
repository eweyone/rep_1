# **Определение формата ссылки для пагинации**

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

titles = []
texts = []

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}

base_url = 'https://lifehacker.ru/topics/technology/?page={}'
```

# **Получение содержимого десяти страниц и ссылок по классам**

```python
article_links = []

for page in range(1, 11):
    url = base_url.format(page)
    response = requests.get(url, headers=headers)

    if response.status_code != 200:
        print(f"Страница {page} не загрузилась, статус {response.status_code}")
        continue

    soup = BeautifulSoup(response.content, 'html.parser')

    articles = soup.find_all('a', class_=['lh-article-card__link', 'lh-small-article-card__link'])
    page_links = ['https://lifehacker.ru' + article['href'] for article in articles]

    print(f"Страница {page}: найдено {len(page_links)} статей")
    article_links.extend(page_links)

    time.sleep(1)

print(f"\n🔗 Всего собрано ссылок: {len(article_links)}")
```

# **Сбор заголовков и текстов статей**

```python
for idx, link in enumerate(article_links, 1):
    response = requests.get(link, headers=headers)

    if response.status_code != 200:
        print(f"Ошибка загрузки статьи {link}")
        titles.append('Ошибка загрузки')
        texts.append('Ошибка загрузки')
        continue

    article_soup = BeautifulSoup(response.content, 'html.parser')

    try:
        title = article_soup.find('h1', class_='article-card__title').get_text(strip=True)
    except AttributeError:
        title = 'Нет заголовка'


    try:
        content_block = article_soup.find('article', id='post-content')
        paragraphs = content_block.find_all('p')
        text = '\n'.join(p.get_text(strip=True) for p in paragraphs)
    except AttributeError:
        text = 'Нет текста'


    titles.append(title)
    texts.append(text)

    print(f"({idx}/{len(article_links)}) {title}")

    time.sleep(1)
```

# **Создание датафрейма**

```python
df = pd.DataFrame({
    'Title': titles,
    'Text': texts
})

print(f"\n✅ Собрано {len(df)} статей.")
df.head()
```
